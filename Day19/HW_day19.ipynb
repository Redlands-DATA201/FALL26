## You Try - Warm-Up 1

Make new data with `make_blobs(n_samples=500, centers=5, cluster_std=1.0, random_state=7)` and pretend you don't know how many groups there are. Make the elbow plot and the silhouette plot. Which $k$ would you choose? Do the two methods agree? Plot the data colored by your clusters.

```python
# Code from lecture - the elbow method on new data
X, y = make_blobs(n_samples=500, centers=5, cluster_std=1.0, random_state=7)

inertias = []
k_values = range(1, 11)
for k in k_values:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X)
    inertias.append(km.inertia_)

plt.plot(k_values, inertias, 'o-')
plt.xlabel("k (number of clusters)")
plt.ylabel("Inertia")
plt.show()
```

## You Try - Warm-Up 2

Run k-Means on the scaled wine data with $k = 2, 3, 4, 5, 6$ and compute the silhouette score for each. Which $k$ does the silhouette score like best? Does it match the number of cultivars? Make the elbow plot too. Is the elbow easy to see with real data?


```python
# Code from lecture - k-Means on the scaled wine data
wine = load_wine(as_frame=True)
X_wine_scaled = StandardScaler().fit_transform(wine.data)

km_wine = KMeans(n_clusters=3, n_init=10, random_state=42)
wine_clusters = km_wine.fit_predict(X_wine_scaled)
print("Silhouette score with k = 3:", silhouette_score(X_wine_scaled, wine_clusters))
```
