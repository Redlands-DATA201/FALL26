```python
# Code from lecture - a decision tree on the made up 3 class data
X, y = make_classification(
    n_samples=200, n_features=2, n_informative=2, n_redundant=0,
    n_clusters_per_class=1,n_classes=3, random_state=42
)
random_seed = 42
# First split: hold out test data
X_train, X_test_full, y_train, y_test_full = train_test_split(
    X, y, test_size=0.2, random_state=random_seed
)
# Second split: create a validation set from the testing data
X_test, X_val, y_test, y_val = train_test_split(
    X_test_full, y_test_full, test_size=0.5, random_state=random_seed
)

clf = DecisionTreeClassifier(max_depth=1,
                             criterion='gini',
                             random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_val)
print(f"Accuracy on the validation data: {accuracy_score(y_val, y_pred):.3f}")

plt.figure(figsize=(12, 6))
plot_tree(clf,
          filled=True,
          feature_names=["Feature 1", "Feature 2"],
          class_names=["Class 0", "Class 1", "Class 2"],
          rounded=True,
          fontsize=10)
plt.show()
```
